```python
import numpy as np
from scipy.linalg import solve_banded
from scipy.linalg import LinAlgError


def part2q1(f, h):
    """
    Part 2, question 1
    Input:
        f: m x n array whose 1st and 2nd derivatives will be computed with respect to x
    Output:
        df, d2f: m x n arrays conaining 1st and 2nd derivatives of f with respect to x
        computed with implicit fd scheme
    """
    # use code below if/as needed
    m, n = f.shape
    h = 1 / (n - 1)
    x = np.linspace(0, 1, n)
    y = np.linspace(0, 1, m)
    df = np.zeros_like(f)  # modify as needed
    d2f = np.zeros_like(f)  # modify as needed

    # Add code here

    # Coefficients for equations
    L1 = [7, h, 16, 0, 7, -h]
    L2 = [-9, -h, 0, 8 * h, 9, -h]
    L1b = [1, 0, 2, -h]
    L2b = [0, h, -6, 5 * h]
    L1b2 = [2, h, 1, 0]
    L2b2 = [-6, -5 * h, 0, -h]

    # Number of equations and band matrix setup
    eqn = 2 * n
    lower = 3
    upper = 3
    ab = np.zeros((lower + upper + 1, eqn))

    # Fill band matrix
    for i in range(n):
        if i == 0:
            # Boundary equations at the left
            for j, val in enumerate(L1b):
                diag = upper + (0 - j)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, j] += val

            for j, val in enumerate(L2b):
                diag = upper + (1 - j)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, j] += val

        elif i == n - 1:
            # Boundary equations at the right
            for j, val in enumerate(L1b2):
                col = eqn - 4 + j
                diag = upper + ((2 * n - 2) - col)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, col] += val

            for j, val in enumerate(L2b2):
                col = eqn - 4 + j
                diag = upper + ((2 * n - 1) - col)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, col] += val

        else:
            # Interior equations
            for j, val in enumerate(L1):
                col = 2 * i - 2 + j
                diag = upper + ((2 * i) - col)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, col] += val

            for j, val in enumerate(L2):
                col = 2 * i - 2 + j
                diag = upper + ((2 * i + 1) - col)
                if 0 <= diag < ab.shape[0]:
                    ab[diag, col] += val

    # RHS coefficients
    c31, c22 = 15 / h, 24 / h
    cb11, cb21, cb31 = -3.5 / h, 4 / h, -0.5 / h
    cb12, cb22, cb32 = 9 / h, -12 / h, 3 / h

    # Initialize derivatives
    df = np.zeros_like(f)
    d2f = np.zeros_like(f)

    # Compute derivatives for each row of f
    for i in range(m):
        b = np.zeros(eqn)
        for j in range(n):
            if j == 0:
                # Left boundary
                b[0] += cb11 * f[i, j] + cb21 * f[i, j + 1] + cb31 * f[i, j + 2]
                b[1] += cb12 * f[i, j] + cb22 * f[i, j + 1] + cb32 * f[i, j + 2]

            elif j == n - 1:
                # Right boundary
                b[2 * j] += -(cb11 * f[i, j] + cb21 * f[i, j - 1] + cb31 * f[i, j - 2])
                b[2 * j + 1] += -(
                    cb12 * f[i, j] + cb22 * f[i, j - 1] + cb32 * f[i, j - 2]
                )

            else:
                # Interior points
                b[2 * j] += c31 * (f[i, j + 1] - f[i, j - 1])
                b[2 * j + 1] += c22 * (f[i, j - 1] - 2 * f[i, j] + f[i, j + 1])

        # Solve the banded linear system
        res_x = solve_banded((lower, upper), ab, b)

        
        df[i, :] = res_x[::2]
        d2f[i, :] = res_x[1::2]

    # Verification with dualfd1
    df_ref = np.zeros_like(f)
    d2f_ref = np.zeros_like(f)
    for i in range(m):
        df_ref[i, :], d2f_ref[i, :] = dualfd1(f[i, :])

    if np.allclose(df, df_ref, atol=1e-8) and np.allclose(d2f, d2f_ref, atol=1e-8):
        print("Verification Successful: Results match within tolerance.")
    else:
        max_diff_df = np.max(np.abs(df - df_ref))
        max_diff_d2f = np.max(np.abs(d2f - d2f_ref))
        print(
            f"Verification Failed: Max difference in df = {max_diff_df}, d2f = {max_diff_d2f}"
        )

    return df, d2f
```
